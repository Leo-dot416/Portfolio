```python
vert_side = int(input("Enter the length of the vertical side in metres:"))
horz_side = int(input("Enter the length of the horizontal side in metres:"))

area_of_object = vert_side * horz_side
print(f"The area of the object is {area_of_object} metres squared.")

pieces = int(input("Enter the number of pieces:"))
price_per_piece = float(input("Enter the price per piece:"))
total_cost = pieces * price_per_piece
print(f"The total cost is £{total_cost}.")
```
